# Energy Efficiency Prediction 🏢⚡
This notebook explores and models the energy efficiency dataset (UCI), focusing on predicting heating and cooling loads.

```
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.preprocessing import StandardScaler
%matplotlib inline
```

## Load the Dataset

```
df = pd.read_csv('data/energy_efficiency_data.csv')
df.head()
```

## Dataset Overview

```
df.info()
```

```
df.describe()
```

## Check for Missing Values

```
df.isnull().sum()
```

## Correlation Matrix

```
plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.title('Feature Correlation')
plt.show()
```

## Standardize the Features

```
X = df.iloc[:, :-2]  # All columns except the last two
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## Heating Load Prediction 🔥

```
y_heating = df['Heating Load']
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y_heating, test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print("Heating Load Prediction")
print(f"MSE: {mean_squared_error(y_test, y_pred):.2f}")
print(f"MAE: {mean_absolute_error(y_test, y_pred):.2f}")
print(f"R²: {r2_score(y_test, y_pred):.2f}")
```

```
plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_pred, alpha=0.7)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--')
plt.xlabel("Actual Heating Load")
plt.ylabel("Predicted Heating Load")
plt.title("Actual vs Predicted Heating Load")
plt.grid(True)
plt.show()
```

## Cooling Load Prediction ❄️

```
y_cooling = df['Cooling Load']
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y_cooling, test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print("Cooling Load Prediction")
print(f"MSE: {mean_squared_error(y_test, y_pred):.2f}")
print(f"MAE: {mean_absolute_error(y_test, y_pred):.2f}")
print(f"R²: {r2_score(y_test, y_pred):.2f}")
```

```
plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_pred, alpha=0.7)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'b--')
plt.xlabel("Actual Cooling Load")
plt.ylabel("Predicted Cooling Load")
plt.title("Actual vs Predicted Cooling Load")
plt.grid(True)
plt.show()
```

## Random Forest Regressor - Heating Load 🌲🔥

```
from sklearn.ensemble import RandomForestRegressor

# Modelo de Random Forest para Heating Load
y_heating = df['Heating Load']
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y_heating, test_size=0.2, random_state=42)

rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)
y_pred_rf = rf_model.predict(X_test)

print("Random Forest - Heating Load")
print(f"MSE: {mean_squared_error(y_test, y_pred_rf):.2f}")
print(f"MAE: {mean_absolute_error(y_test, y_pred_rf):.2f}")
print(f"R²: {r2_score(y_test, y_pred_rf):.2f}")
```

```
plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_pred_rf, alpha=0.7)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'g--')
plt.xlabel("Actual Heating Load")
plt.ylabel("Predicted Heating Load (RF)")
plt.title("Random Forest: Actual vs Predicted Heating Load")
plt.grid(True)
plt.show()
```
